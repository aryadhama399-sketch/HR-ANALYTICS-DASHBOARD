# HR Analytics Dashboard

Data cleaning, EDA, attrition analysis and business insights.

```
import pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\nfrom pathlib import Path
```

```
df = pd.read_csv('../data/HR_Employee_Attrition_Cleaned.csv')\ndf.shape
```

```
df.info()\ndf.isna().sum().sum()
```

## KPI Analysis

```
total_employees = len(df)
attrition_count = (df['Attrition']=='Yes').sum()
attrition_rate = attrition_count/total_employees*100
print(f'Total employees: {total_employees:,}')
print(f'Attrition count: {attrition_count:,}')
print(f'Attrition rate: {attrition_rate:.2f}%')
print(f'Average age: {df.Age.mean():.1f}')
print(f'Average monthly income: ${df.MonthlyIncome.mean():,.0f}')
print(f'Average years at company: {df.YearsAtCompany.mean():.1f}')
```

## Department Attrition

```
dept = df.groupby('Department').agg(
    Employees=('EmployeeNumber','count'),
    Attrition_Count=('Attrition', lambda x:(x=='Yes').sum())
)
dept['Attrition_Rate_%'] = dept['Attrition_Count']/dept['Employees']*100
dept.sort_values('Attrition_Rate_%', ascending=False)
```

## Overtime Analysis

```
df.groupby('OverTime')['Attrition'].apply(lambda x:(x=='Yes').mean()*100).sort_values(ascending=False)
```

## Job Role Analysis

```
role = df.groupby('JobRole').agg(
    Employees=('EmployeeNumber','count'),
    Attrition_Count=('Attrition', lambda x:(x=='Yes').sum()),
    Avg_Monthly_Income=('MonthlyIncome','mean')
)
role['Attrition_Rate_%'] = role['Attrition_Count']/role['Employees']*100
role.sort_values('Attrition_Rate_%', ascending=False)
```
